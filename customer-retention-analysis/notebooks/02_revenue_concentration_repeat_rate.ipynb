# Revenue concentration.
### Business kitne customers par tika hai?

```python
cust_summary = cust_summary.sort_values(by='total_revenue',ascending=False)
```

```python
total_revenue_sum = cust_summary['total_revenue'].sum()
total_revenue_sum
```

```python
n = int(cust_summary['customer_id'].count()*.2)
n
```

```python
top20_revenue = cust_summary.head(n)['total_revenue'].sum()
```

```python
shareof_top20 = (top20_revenue / total_revenue_sum) * 100
```

```python
shareof_top20
```

```python
top10_revenue = cust_summary.head(int(cust_summary['customer_id'].count()*.1))['total_revenue'].sum()
shareof_top10 = (top10_revenue / total_revenue_sum) * 100
```

```python
print(f"top 10 % :- {round(shareof_top10)}  top 20 %:- {round(shareof_top20)}")
```

```python
top50_revenue = cust_summary.head(int(cust_summary['customer_id'].count()*.5))['total_revenue'].sum()
shareof_top50 = (top50_revenue / total_revenue_sum) * 100

print(round(shareof_top50))
```

```python
top80_revenue = cust_summary.head(int(cust_summary['customer_id'].count()*.8))['total_revenue'].sum()
top80_cust = cust_summary.head(int(cust_summary['customer_id'].count() * .8))['customer_id'].count()
print(top80_cust , top80_revenue)
```

### "Top 20% customers 62% revenue laate hain, isliye in customers ko khona sabse mehnga padega."

## sirf top 5% customers ko sambhalna kaafi nahi hoga, retention ka kaam ek bade segment par karna padega.

```python
cust_summary['cumulative revenue'] = cust_summary['total_revenue'].cumsum()

cust_summary['cum%'] = cust_summary['cumulative revenue'] / total_revenue_sum
customer_80_count = (cust_summary['cum%'] < 0.8).sum() + 1


print ( f"customer pecentage that contribute 80% in revenue is:- {int((customer_80_count/cust_summary['customer_id'].count())*100)}%")
```
