# MASTER TABLE

```python
cleaned_customers = pd.read_csv(r'C:\Users\ASUS\OneDrive\Desktop\izaan_coding\project\customer-retention-analysis\data\processed\cleaned_customers.csv')
```

```python
cleaned_orders = pd.read_csv(r'C:\Users\ASUS\OneDrive\Desktop\izaan_coding\project\customer-retention-analysis\data\processed\cleaned_orders.csv')
cleaned_order_items = pd.read_csv(r'C:/Users/ASUS/OneDrive/Desktop/izaan_coding/project/customer-retention-analysis/data/processed/cleaned_order_items.csv')
cleaned_products = pd.read_csv(r'C:/Users/ASUS/OneDrive/Desktop/izaan_coding/project/customer-retention-analysis/data/processed/cleaned_products.csv')
```

```python
print(cleaned_orders['order_id'].is_unique)       
print(cleaned_customers['customer_id'].is_unique)  
print(cleaned_products['product_id'].is_unique)  
```

```python
cleaned_orders['order_date'] = pd.to_datetime(cleaned_orders['order_date'])
cleaned_customers['signup_date'] = pd.to_datetime(cleaned_customers['signup_date'])
```

```python
n = len(cleaned_order_items)

master = cleaned_order_items.merge(cleaned_orders, on='order_id', how='left')
assert len(master) == n, "orders join me rows badh gayi"

master = master.merge(cleaned_products, on='product_id', how='left')
assert len(master) == n, "products join me rows badh gayi"

master = master.merge(cleaned_customers, on='customer_id', how='left')
assert len(master) == n, "customers join me rows badh gayi"

```

```python
print(master['customer_id'].isna().sum(),
      master['category'].isna().sum(),
      master['signup_date'].isna().sum())
```

```python
master['line_revenue'] = master['quantity'] * master['unit_price'] - master['discount_amount']
master['order_month'] = master['order_date'].dt.to_period('M')
```

```python
sales = master[master['status'] == 'completed'].copy()
```

```python
print(len(master), len(sales))
print(round(sales['line_revenue'].sum(), 2))
```
