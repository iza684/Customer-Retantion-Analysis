## RFM segmentation

```python
cust_summary['first_order'] = pd.to_datetime(cust_summary['first_order'])
cust_summary['last_order'] = pd.to_datetime(cust_summary['last_order'])
```

```python
snapshot_date = sales['order_date'].max() +pd.Timedelta(days=1)
snapshot_date
```

```python
recency = (snapshot_date - cust_summary['last_order']).dt.days
recency
```

```python
cust_summary['recency'] = recency
```

```python
cust_summary[['recency', 'num_orders', 'total_revenue']].describe()
```

```python
labels=[4,3,2,1]
```

```python
r_score = pd.qcut(recency,q=4,labels=labels)
r_score
```

```python
f_score = pd.cut(
    cust_summary['num_orders'],
    bins=[0, 1, 2, 4, np.inf],
    labels=[1, 2, 3, 4]
).astype(int)
f_score
```

```python
m_score = pd.qcut(cust_summary['total_revenue'],q=4,labels=[1,2,3,4])
m_score
```

```python
cust_summary['f_score'] = f_score
```

```python
cust_summary['m_score'] = m_score
cust_summary['r_score'] = r_score
```

```python
r = cust_summary['r_score'].astype(int)
f = cust_summary['f_score'].astype(int)
m = cust_summary['m_score'].astype(int)

conditions = [
    (r == 4) & (f >= 3) & (m >= 3),
    (r >= 3) & (f >= 2),
    (r >= 3) & (f == 1),
    (r <= 2) & (f >= 2),
    (r <= 2) & (f == 1)
]

labels = ["Champions", "Loyal", "New", "At-Risk", "Lost"]
cust_summary['segment'] = np.select(conditions, labels, default='Unassigned')
```

```python
cust_summary['segment'].isna().sum()
```

```python
print(cust_summary['segment'].value_counts())
print(cust_summary['segment'].isna().sum())
```

```python
 v=cust_summary.groupby('segment').agg(
    total_cus = ('customer_id','count'),
    total_rev = ('total_revenue','sum'),
    avg_rev = ('total_revenue','mean'), 
    avg_recency = ('recency','mean'),
    avg_num_orders = ('num_orders','mean')
 )  
v['revenue_pct'] = round((v['total_rev'] / v['total_rev'].sum()) * 100)
v
```
